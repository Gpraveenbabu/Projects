```python
movies = pd.read_csv(
    "tmdb_5000_movies.csv"
)

credits = pd.read_csv(
    "tmdb_5000_credits.csv"
)
anime = pd.read_csv(
    "anime-dataset-2023.csv"
)
```

```python
movies = movies[
    [
        'title',
        'genres',
        'overview',
        'vote_average',
        'release_date'
    ]
]

credits = credits[
    [
        'title',
        'cast',
        'crew'
    ]
]

movies = movies.merge(
    credits,
    on='title'
)
```

```python
anime = anime[
    [
        'Name',
        'Genres',
        'Score',
        'Synopsis',
        'Studios',
        'Type'
    ]
]

anime.rename(columns={
    'Name': 'title',
    'Genres': 'genres',
    'Score': 'vote_average',
    'Synopsis': 'overview',
    'Studios': 'cast',
    'Type': 'crew'
}, inplace=True)

anime['release_date'] = ''

anime['year'] = ''
```

```python
imdb_data = pd.read_csv(
    "movies.csv"
)
```

```python
imdb_data = imdb_data[
    [
        'name',
        'genre',
        'rating',
        'year'
    ]
]

imdb_data.rename(columns={
    'name': 'title',
    'genre': 'genres',
    'rating': 'vote_average'
}, inplace=True)

imdb_data['overview'] = ''

imdb_data['cast'] = ''

imdb_data['crew'] = ''

imdb_data['release_date'] = ''
```

```python
combined = pd.concat(
    [
        movies,
        anime,
        imdb_data
    ],
    ignore_index=True
)
```

```python
combined['tags'] = (
    combined['title'].astype(str) + ' ' +
    combined['genres'].astype(str) + ' ' +
    combined['overview'].astype(str) + ' ' +
    combined['cast'].astype(str) + ' ' +
    combined['crew'].astype(str) + ' ' +
    combined['year'].astype(str)
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
tfidf = TfidfVectorizer(
    stop_words='english',
    max_features=30000
)

vectors = tfidf.fit_transform(
    combined['tags']
)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def recommend(query, top_n=10):

    query = query.lower()

    matches = combined[
        combined['tags']
        .str.lower()
        .str.contains(query, case=False, na=False)
    ]

    if matches.empty:
        print("No results found")
        return

    idx = matches.index[0]

    # Compute similarity ONLY for one movie
    distances = cosine_similarity(
        vectors[idx],
        vectors
    ).flatten()

    recommendations = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:100]

    final = []

    for i in recommendations:

        row = combined.iloc[i[0]]

        try:
            rating = float(row['vote_average'])
        except:
            rating = 0

        final.append({
            "title": row['title'],
            "similarity": i[1],
            "rating": rating
        })

    final = sorted(
        final,
        key=lambda x: (
            x['similarity'],
            x['rating']
        ),
        reverse=True
    )

    print("\nTop Recommendations:\n")

    for item in final[:top_n]:

        print(
            f"{item['title']} "
            f"| Similarity: {round(item['similarity'],3)} "
            f"| Rating: {item['rating']}"
        )
```

```python
def recommend(query, top_n=10):

    query = query.lower()

    matches = combined[
        combined['tags']
        .str.lower()
        .str.contains(query, case=False, na=False)
    ]

    if matches.empty:
        print("No results found")
        return

    idx = matches.index[0]

    # FAST SIMILARITY
    distances = cosine_similarity(
        vectors[idx],
        vectors
    ).flatten()

    recommendations = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:100]

    final = []

    for i in recommendations:

        row = combined.iloc[i[0]]

        try:
            rating = float(row['vote_average'])
        except:
            rating = 0

        final.append({
            "title": row['title'],
            "similarity": i[1],
            "rating": rating
        })

    # SORT BY SIMILARITY + RATING
    final = sorted(
        final,
        key=lambda x: (
            x['similarity'],
            x['rating']
        ),
        reverse=True
    )

    print("\nTop Recommendations:\n")

    shown = set()

    for item in final:

        # REMOVE DUPLICATES
        if item['title'] not in shown:

            shown.add(item['title'])

            print(
                f"🎥 {item['title']} "
                f"| Similarity: {round(item['similarity'],3)} "
                f"| Rating: {item['rating']}"
            )

        if len(shown) >= top_n:
            break

# =========================
# USER INPUT
# =========================

query = input(
    "Enter movie, anime, actor, director, genre, or year: "
)

recommend(query)
```
